##  DATA CLEANING & PREPARATION

### CREATE A BACKUP OF RAW DATA

```python

df_raw = df.copy()
```

#### 1. STANDARDIZE COLUMN NAMES

```python
df.columns = (
    df.columns
    .str.strip()
    .str.replace(" ", "_")
)

print("\nColumn names standardized.")
```

#### 2.REMOVE EXTRA SPACES FROM TEXT COLUMNS

```python
text_columns = df.select_dtypes(
    include=["object", "str"]
).columns

for col in text_columns:
    df[col] = df[col].apply(
        lambda x: x.strip() if isinstance(x, str) else x
    )

print("Text spaces cleaned successfully.")
```

#### 3.CHECK DUPLICATE ROWS

```python
duplicate_count = df.duplicated().sum()

print("\nDuplicate rows found:", duplicate_count)

if duplicate_count > 0:
    df = df.drop_duplicates()
    print("Duplicate rows removed.")
else:
    print("No duplicate rows found.")
```

#### 4.CONVERT DATE COLUMNS

```python
date_columns = [
    "Customer_Since",
    "Due_Date",
    "Loan_Start_Date",
    "Loan_End_Date",
    "Payment_Date",
    "Transaction_Date"
]

for col in date_columns:

    if col in df.columns:

        df[col] = pd.to_datetime(
            df[col],
            errors="coerce"
        )

print("\nDate columns converted.")
```

#### 5.DATE COVERSION 

```python
print("\nDate columns:")
print(df[date_columns].dtypes)
```

#### 6.HANDLE CATEGORICAL MISSING VALUES

```python
categorical_missing_columns = [
    "Payment_Status",
    "Delinquency_Status",
    "Employment_Status",
    "Customer_Segment",
    "Occupation",
    "Country",
    "City",
    "Gender",
    "Collateral_Type",
    "Transaction_Type",
    "Loan_Type",
    "Loan_Status",
    "Transaction_Status",
    "Transaction_Channel",
    "Merchant_Category"
]
for col in categorical_missing_columns:

    if col in df.columns:
        df[col] = df[col].fillna("Unknown")

print("\nCategorical missing values handled.")
```

#### 7.IDENTIFIER COLUMNS

```python

identifier_columns = [
    "Customer_ID",
    "Loan_ID",
    "Payment_ID",
    "Transaction_ID"
]

print("\nIdentifier columns checked.")
```

#### 8.NUMERIC MISSING VALUE HANDLING

```python
numeric_columns = df.select_dtypes(
    include=["int64", "float64"]
).columns.tolist()

print("\nNumeric columns:")
print(numeric_columns)
```

#### 9.FIX BOOLEAN/FLAG COLUMNS

```python
flag_columns = [
    "Missed_Payment_Flag",
    "Default_Flag",
    "Early_Warning_Flag"
]

for col in flag_columns:

    if col in df.columns:

        # Convert to numeric
        df[col] = pd.to_numeric(
            df[col],
            errors="coerce"
        )

        # Keep only valid binary values
        df.loc[
            ~df[col].isin([0, 1]) & df[col].notna(),
            col
        ] = np.nan

print("\nFlag columns validated.")
```

#### 10.VALIDATE AGE

```python
if "Age" in df.columns:

    invalid_age = (
        (df["Age"] < 18) |
        (df["Age"] > 100)
    )

    print(
        "\nInvalid ages:",
        invalid_age.sum()
    )

    df.loc[invalid_age, "Age"] = np.nan
```

#### 11.VALIDATE CREDIT SCORE

```python
if "Credit_Score" in df.columns:

    invalid_credit_score = (
        (df["Credit_Score"] < 300) |
        (df["Credit_Score"] > 850)
    )

    print(
        "Invalid credit scores:",
        invalid_credit_score.sum()
    )

    df.loc[
        invalid_credit_score,
        "Credit_Score"
    ] = np.nan
```

#### 12.VALIDATE INTEREST RATE

```python
if "Interest_Rate" in df.columns:

    invalid_interest_rate = (
        (df["Interest_Rate"] < 0) |
        (df["Interest_Rate"] > 100)
    )

    print(
        "Invalid interest rates:",
        invalid_interest_rate.sum()
    )

    df.loc[
        invalid_interest_rate,
        "Interest_Rate"
    ] = np.nan
```

#### 13.VALIDATE UTILIZATION RATE

```python
if "Utilization_Rate" in df.columns:

    invalid_utilization = (
        (df["Utilization_Rate"] < 0) |
        (df["Utilization_Rate"] > 100)
    )

    print(
        "Invalid utilization rates:",
        invalid_utilization.sum()
    )

    df.loc[
        invalid_utilization,
        "Utilization_Rate"
    ] = np.nan
```

#### 14.VALIDATE DEBT TO INCOME RATIO

```python
 if "Debt_to_Income_Ratio" in df.columns:

    invalid_dti = (
        df["Debt_to_Income_Ratio"] < 0
    )

    print(
        "Invalid DTI values:",
        invalid_dti.sum()
    )

    df.loc[
        invalid_dti,
        "Debt_to_Income_Ratio"
    ] = np.nan
```

#### 15.VALIDATE FINANCIAL AMOUNTS

```python
financial_columns = [
    "Annual_Income",
    "Transaction_Amount",
    "Loan_Amount",
    "Payment_Amount",
    "Amount_Paid",
    "Monthly_Installment",
    "Account_Balance",
    "Interest_Income",
    "Fee_Income",
    "Transaction_Fees",
    "Service_Cost",
    "Customer_Profit",
    "Credit_Loss"
]

for col in financial_columns:
     if col in df.columns:

        negative_count = (
            df[col].dropna() < 0
        ).sum()

        print(
            f"{col} - negative values:",
            negative_count
        )
```

#### 16.VALIDATE MISSED PAYMENT COUNT

```python
if "Missed_Payment_Count" in df.columns:

    invalid_missed_payment = (
        df["Missed_Payment_Count"] < 0
    )

    print(
        "\nInvalid missed payment counts:",
        invalid_missed_payment.sum()
    )

    df.loc[
        invalid_missed_payment,
        "Missed_Payment_Count"
    ] = np.nan
```

#### 17. VALIDATE DAYS PAST DUE

```python
dpd_columns = [
    "Days_Past_Due_x",
    "Days_Past_Due_y"
]

for col in dpd_columns:

    if col in df.columns:

        invalid_dpd = (
            df[col] < 0
        )

        print(
            f"{col} invalid values:",
            invalid_dpd.sum()
        )

        df.loc[
            invalid_dpd,
            col
        ] = np.nan
```

#### 18.CHECK THE TWO DAYS PAST DUE COLUMNS

```python
if (
    "Days_Past_Due_x" in df.columns
    and
    "Days_Past_Due_y" in df.columns
):

    comparison = pd.DataFrame({
        "Days_Past_Due_x": df["Days_Past_Due_x"],
        "Days_Past_Due_y": df["Days_Past_Due_y"]
    })

    comparison["Difference"] = (
        comparison["Days_Past_Due_x"]
        -
        comparison["Days_Past_Due_y"]
    )

    print("\nDays Past Due comparison:")
    print(
        comparison["Difference"]
        .value_counts(dropna=False)
        .head(15)
    )
```

#### 19.CREATE A SINGLE DAYS PAST DUE COLUMN

```python
# We don't delete the original columns yet.
# First create a consolidated field.

if (
    "Days_Past_Due_x" in df.columns
    and
    "Days_Past_Due_y" in df.columns
):

    df["Days_Past_Due"] = (
        df["Days_Past_Due_y"]
        .combine_first(df["Days_Past_Due_x"])
    )

print("\nConsolidated Days_Past_Due column created.")
```

#### 20.CREATE AGE GROUP

```python
if "Age" in df.columns:

    df["Age_Group"] = pd.cut(
        df["Age"],
        bins=[17, 25, 35, 50, 65, 100],
        labels=[
            "adult",
            "mid-adult",
            "mid-age",
            "old_age",
            "old"
        ]
    )

print("\nAge_Group created.")
```

#### 21.CREATE PROFIT MARGIN CHECK

```python
# We won't overwrite the original Profit_Margin.
# We'll create a validation column.

if (
    "Customer_Profit" in df.columns
    and
    "Annual_Income" in df.columns
):

    df["Calculated_Profit_Margin"] = np.where(
        df["Annual_Income"] != 0,
        (df["Customer_Profit"] /
         df["Annual_Income"]) * 100,
        np.nan
    )

print("\nCalculated profit margin check created.")
```

#### 22.CHECK CUSTOMER ID / CUSTOMER NAME CONSISTENCY

```python
if (
    "Customer_ID" in df.columns
    and
    "Customer_Name" in df.columns
):

    customer_mapping = (
        df[
            ["Customer_ID", "Customer_Name"]
        ]
        .drop_duplicates()
    )

    customer_id_name_counts = (
        customer_mapping
        .groupby("Customer_ID")
        ["Customer_Name"]
        .nunique()
    )

    inconsistent_customers = (
        customer_id_name_counts > 1
    ).sum()

    print(
        "\nCustomer IDs with multiple names:",
        inconsistent_customers
    )
```

#### 23.CHECK LOAN STATUS LOGIC

```python
if (
    "Loan_Status" in df.columns
    and
    "Default_Flag" in df.columns
):

    loan_status_default_check = pd.crosstab(
        df["Loan_Status"],
        df["Default_Flag"],
        dropna=False
    )

    print("\nLoan Status vs Default Flag:")
    print(loan_status_default_check)
```

#### 24.CHECK PAYMENT STATUS LOGIC

```python
if (
    "Payment_Status" in df.columns
    and
    "Amount_Paid" in df.columns
):

    payment_status_summary = (
        df.groupby("Payment_Status")[
            "Amount_Paid"
        ]
        .agg(
            ["count", "sum", "mean"]
        )
    )

    print("\nPayment status summary:")
    print(payment_status_summary)
```

#### 25.CHECK TRANSACTION STATUS

```python
if (
    "Transaction_Status" in df.columns
    and
    "Transaction_Amount" in df.columns
):

    transaction_status_summary = (
        df.groupby("Transaction_Status")[
            "Transaction_Amount"
        ]
        .agg(
            ["count", "sum", "mean"]
        )
    )

    print("\nTransaction status summary:")
    print(transaction_status_summary)
```

#### 26. FINAL MISSING VALE REPORT 

```python
final_missing = pd.DataFrame({
    "Column": df.columns,
    "Missing_Count": df.isna().sum(),
    "Missing_Percentage": (
        df.isna().sum()
        /
        len(df)
        *
        100
    ).round(2)
})

final_missing = final_missing.sort_values(
    "Missing_Percentage",
    ascending=False
)

print("\n" + "=" * 60)
print("FINAL MISSING VALUE REPORT")
print("=" * 60)

print(final_missing)
```

## FINAL CLEANING

```python
# ============================================================
#### 1. REMOVE OLD / INCORRECT PROFIT MARGIN COLUMN
# ============================================================

if "Calculated_Profit_Margin" in df.columns:

    df = df.drop(
        columns=["Calculated_Profit_Margin"]
    )

    print(
        "\nRemoved old Calculated_Profit_Margin column."
    )


# ============================================================
#### 2. INVESTIGATE DAYS PAST DUE COLUMNS
# ============================================================

print("\n" + "=" * 70)
print("DAYS PAST DUE INVESTIGATION")
print("=" * 70)

if (
    "Days_Past_Due_x" in df.columns
    and
    "Days_Past_Due_y" in df.columns
):

    dpd_compare = pd.DataFrame({
        "x": df["Days_Past_Due_x"],
        "y": df["Days_Past_Due_y"]
    })

    dpd_compare["difference"] = (
        dpd_compare["x"]
        -
        dpd_compare["y"]
    )

    print("\nNon-matching DPD records:")

    print(
        (
            dpd_compare["difference"]
            .notna()
            &
            (dpd_compare["difference"] != 0)
        ).sum()
    )

    print("\nDPD comparison:")
    print(
        dpd_compare["difference"]
        .value_counts(dropna=False)
        .head(15)
    )


# ============================================================
#### 3. USE CONSOLIDATED DAYS PAST DUE
# ============================================================

if (
    "Days_Past_Due" not in df.columns
    and
    "Days_Past_Due_y" in df.columns
    and
    "Days_Past_Due_x" in df.columns
):

    df["Days_Past_Due"] = (
        df["Days_Past_Due_y"]
        .combine_first(
            df["Days_Past_Due_x"]
        )
    )

print(
    "\nConsolidated Days_Past_Due available:",
    "Days_Past_Due" in df.columns
)


# ============================================================
#### 4. CREATE DPD RISK BAND
# ============================================================

if "Days_Past_Due" in df.columns:

    df["DPD_Risk_Band"] = pd.cut(
        df["Days_Past_Due"],
        bins=[
            -1,
            0,
            30,
            60,
            90,
            np.inf
        ],
        labels=[
            "Current",
            "1-30 DPD",
            "31-60 DPD",
            "61-90 DPD",
            "90+ DPD"
        ]
    )

print("\nDPD_Risk_Band created.")


# ============================================================
#### 5. CREATE CREDIT SCORE BAND
# ============================================================

if "Credit_Score" in df.columns:

    df["Credit_Score_Band"] = pd.cut(
        df["Credit_Score"],
        bins=[
            299,
            579,
            669,
            739,
            799,
            850
        ],
        labels=[
            "Poor",
            "Fair",
            "Good",
            "Very Good",
            "Excellent"
        ]
    )

print("Credit_Score_Band created.")


# ============================================================
#### 6. CREATE RISK SCORE BAND
# ============================================================

if "Risk_Score" in df.columns:

    df["Risk_Score_Band"] = pd.cut(
        df["Risk_Score"],
        bins=[
            -np.inf,
            59,
            69,
            79,
            np.inf
        ],
        labels=[
            "Low",
            "Moderate",
            "High",
            "Very High"
        ]
    )

print("Risk_Score_Band created.")


# ============================================================
#### 7. CREATE UTILIZATION BAND
# ============================================================

if "Utilization_Rate" in df.columns:

    df["Utilization_Band"] = pd.cut(
        df["Utilization_Rate"],
        bins=[
            -0.01,
            30,
            50,
            75,
            100
        ],
        labels=[
            "Low",
            "Moderate",
            "High",
            "Very High"
        ]
    )

print("Utilization_Band created.")


# ============================================================
#### 8. CREATE DTI BAND
# ============================================================

if "Debt_to_Income_Ratio" in df.columns:

    df["DTI_Band"] = pd.cut(
        df["Debt_to_Income_Ratio"],
        bins=[
            -0.01,
            20,
            36,
            50,
            np.inf
        ],
        labels=[
            "Low",
            "Moderate",
            "High",
            "Very High"
        ]
    )

print("DTI_Band created.")


# ============================================================
#### 9. CREATE PAYMENT BEHAVIOR CATEGORY
# ============================================================

if "Payment_Status" in df.columns:

    df["Payment_Behavior"] = np.select(
        [
            df["Payment_Status"].eq("Paid"),
            df["Payment_Status"].eq("Partial"),
            df["Payment_Status"].eq("Missed"),
            df["Payment_Status"].eq("Unknown")
        ],
        [
            "On Time / Paid",
            "Partial Payment",
            "Missed Payment",
            "No Payment Record"
        ],
        default="Other"
    )

print("Payment_Behavior created.")


# ============================================================
#### 10. CREATE DEFAULT STATUS
# ============================================================

if "Default_Flag" in df.columns:

    df["Default_Status"] = np.select(
        [
            df["Default_Flag"].eq(1),
            df["Default_Flag"].eq(0)
        ],
        [
            "Default",
            "No Default"
        ],
        default="No Payment Record"
    )

print("Default_Status created.")


# ============================================================
#### 11. CREATE EARLY WARNING STATUS
# ============================================================

if "Early_Warning_Flag" in df.columns:

    df["Early_Warning_Status"] = np.select(
        [
            df["Early_Warning_Flag"].eq(1),
            df["Early_Warning_Flag"].eq(0)
        ],
        [
            "Early Warning",
            "No Early Warning"
        ],
        default="Unknown"
    )

print("Early_Warning_Status created.")


# ============================================================
#### 12. CREATE LOAN DURATION
# ============================================================

if (
    "Loan_Start_Date" in df.columns
    and
    "Loan_End_Date" in df.columns
):

    df["Loan_Duration_Days"] = (
        df["Loan_End_Date"]
        -
        df["Loan_Start_Date"]
    ).dt.days

    print(
        "\nLoan duration created."
    )


# ============================================================
#### 13. CREATE LOAN TERM BAND
# ============================================================

if "Loan_Term_Months" in df.columns:

    df["Loan_Term_Band"] = pd.cut(
        df["Loan_Term_Months"],
        bins=[
            -1,
            12,
            24,
            36,
            60,
            np.inf
        ],
        labels=[
            "0-12 Months",
            "13-24 Months",
            "25-36 Months",
            "37-60 Months",
            "60+ Months"
        ]
    )

print("Loan_Term_Band created.")


# ============================================================
#### 14. CREATE PROFITABILITY STATUS
# ============================================================

if "Customer_Profit" in df.columns:

    df["Profitability_Status"] = np.select(
        [
            df["Customer_Profit"] > 0,
            df["Customer_Profit"] < 0,
            df["Customer_Profit"].eq(0)
        ],
        [
            "Profitable",
            "Loss-Making",
            "Break-Even"
        ],
        default="Unknown"
    )

print("Profitability_Status created.")


# ============================================================
#### 15. CREATE CREDIT LOSS STATUS
# ============================================================

if "Credit_Loss" in df.columns:

    df["Credit_Loss_Status"] = np.select(
        [
            df["Credit_Loss"] > 0,
            df["Credit_Loss"].eq(0)
        ],
        [
            "Credit Loss",
            "No Credit Loss"
        ],
        default="Unknown"
    )

print("Credit_Loss_Status created.")


# ============================================================
#### 16. CREATE CUSTOMER RISK INDICATOR
# ============================================================

risk_conditions = []

if "Default_Flag" in df.columns:
    risk_conditions.append(
        df["Default_Flag"].eq(1)
    )

if "Early_Warning_Flag" in df.columns:
    risk_conditions.append(
        df["Early_Warning_Flag"].eq(1)
    )

if "Days_Past_Due" in df.columns:
    risk_conditions.append(
        df["Days_Past_Due"] > 30
    )

if risk_conditions:

    df["High_Risk_Indicator"] = np.where(
        np.logical_or.reduce(
            risk_conditions
        ),
        1,
        0
    )

    print(
        "High_Risk_Indicator created."
    )


# ============================================================
#### 17. VALIDATE NEW FEATURES
# ============================================================

new_columns = [
    "DPD_Risk_Band",
    "Credit_Score_Band",
    "Risk_Score_Band",
    "Utilization_Band",
    "DTI_Band",
    "Payment_Behavior",
    "Default_Status",
    "Early_Warning_Status",
    "Loan_Duration_Days",
    "Loan_Term_Band",
    "Profitability_Status",
    "Credit_Loss_Status",
    "High_Risk_Indicator"
]

print("\n" + "=" * 70)
print("NEW FEATURE VALIDATION")
print("=" * 70)

for col in new_columns:

    if col in df.columns:

        print(
            f"\n{col}:"
        )

        print(
            df[col]
            .value_counts(
                dropna=False
            )
            .head(15)
        )


# ============================================================
#### 18. CHECK NEGATIVE FINANCIAL VALUES
# ============================================================

print("\n" + "=" * 70)
print("NEGATIVE FINANCIAL VALUE CHECK")
print("=" * 70)

for col in financial_columns:

    if col in df.columns:

        negative_count = (
            df[col].dropna() < 0
        ).sum()

        print(
            f"{col}: {negative_count}"
        )


# ============================================================
#### 19. CHECK INVALID LOAN DURATIONS
# ============================================================

if "Loan_Duration_Days" in df.columns:

    invalid_duration = (
        df["Loan_Duration_Days"] < 0
    ).sum()

    print(
        "\nInvalid loan durations:",
        invalid_duration
    )


# ============================================================
#### 20. REMOVE REDUNDANT DPD COLUMNS
# ============================================================

if (
    "Days_Past_Due_x" in df.columns
    and
    "Days_Past_Due_y" in df.columns
):

    difference_count = (
        (
            df["Days_Past_Due_x"]
            -
            df["Days_Past_Due_y"]
        )
        .abs()
        .gt(0)
        .sum()
    )

    print(
        "\nDifferent x/y DPD records:",
        difference_count
    )

    if difference_count == 0:

        df = df.drop(
            columns=[
                "Days_Past_Due_x",
                "Days_Past_Due_y"
            ]
        )

        print(
            "Original duplicate DPD columns removed."
        )

    else:

        print(
            "DPD columns retained because differences exist."
        )


# ============================================================
#### 21. FINAL COLUMN CHECK
# ============================================================

print("\n" + "=" * 70)
print("FINAL COLUMN LIST")
print("=" * 70)

for i, col in enumerate(
    df.columns,
    start=1
):

    print(
        i,
        col
    )


# ============================================================
#### 22. FINAL MISSING VALUE REPORT
# ============================================================

final_missing = pd.DataFrame({
    "Column": df.columns,
    "Missing_Count": df.isna().sum(),
    "Missing_Percentage": (
        df.isna().sum()
        /
        len(df)
        *
        100
    ).round(2)
})

final_missing = final_missing.sort_values(
    "Missing_Percentage",
    ascending=False
)

print("\n" + "=" * 70)
print("FINAL MISSING VALUE REPORT")
print("=" * 70)

print(final_missing)


# ============================================================
#### 23. FINAL DATASET SHAPE
# ============================================================

print("\n" + "=" * 70)
print("FINAL DATASET SHAPE")
print("=" * 70)

print(
    "Rows:",
    df.shape[0]
)

print(
    "Columns:",
    df.shape[1]
)


# ============================================================
#### 24. SAVE FINAL PYTHON DATASET
# ============================================================

final_output = (
    "Fintech_Banking_Final.csv"
)

df.to_csv(
    final_output,
    index=False
)

print(
    "\nFinal dataset saved as:"
)

print(
    final_output
)


# ============================================================
#### 25. SAVE FINAL DATA QUALITY REPORT
# ============================================================

final_missing.to_csv(
    "04_final_data_quality_report.csv",
    index=False
)

print(
    "Final quality report saved as:"
)

print(
    "04_final_data_quality_report.csv"
)


# ============================================================
#### 26. FINAL CONFIRMATION
# ============================================================

print("\n" + "=" * 70)
print("STEP 4 COMPLETED SUCCESSFULLY")
print("=" * 70)

print(
    "Final dataset:",
    df.shape
)

print(
    "Ready for SQL database preparation."
)
```
